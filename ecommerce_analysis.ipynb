```python
df = orders.merge(
    order_item,
    on="order_id",
    how="inner"
)
```

```python
df.head()
df.shape
```

```python
df = df.merge(
    products,
    on="product_id",
    how="inner"
)
```

```python
df = df.merge(
    customers,
    on="customer_id",
    how="inner"
)
```

```python
df = df.merge(
    payments,
    on="order_id",
    how="left"
)
```

```python
df.head()
df.shape
```

```python
df.columns
```

```python
df["revenue"] = (
    df["price"]
    * df["quantity"]
    * (1 - df["discount"] / 100)
)
```

```python
df[
    [
        "product_name",
        "price",
        "quantity",
        "discount",
        "revenue"
    ]
].head(10)
```

```python
df["cost"] = (
    df["cost_price"]
    * df["quantity"]
)
```

```python
df["profit"] = (
    df["revenue"]
    - df["cost"]
)
```

```python
df[
    [
        "product_name",
        "revenue",
        "cost",
        "profit"
    ]
].head(10)
```

```python
df["profit_margin"] = np.where(
    df["revenue"] != 0,
    (df["profit"] / df["revenue"]) * 100,
    0
)
```

```python
df["year"] = df["order_date"].dt.year
```

```python
df["month"] = df["order_date"].dt.month
```

```python
df["month_name"] = df["order_date"].dt.month_name()
```

```python
df["day"] = df["order_date"].dt.day
```

```python
df["day_of_week"] = df["order_date"].dt.day_name()
```
